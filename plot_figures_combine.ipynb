```python
import matplotlib.pyplot as plt
import fitz  # PyMuPDF
import numpy as np
from PIL import Image

# 将PDF页面转换为PIL图像，增加渲染分辨率
def pdf_to_image(pdf_path, zoom=2):
    # 使用 PyMuPDF 打开PDF文件
    pdf_document = fitz.open(pdf_path)
    
    # 获取第一页并渲染为高分辨率图像
    page = pdf_document.load_page(0)
    
    # 设置渲染矩阵，zoom控制DPI，值越大分辨率越高
    mat = fitz.Matrix(zoom, zoom)  # 放大2倍，提升分辨率
    pix = page.get_pixmap(matrix=mat)
    
    # 将图像保存为PIL Image
    img = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)
    pdf_document.close()
    return img

# 读取PDF图片，并增加分辨率
img1 = pdf_to_image('IMFs.pdf')
img2 = pdf_to_image('QPO_decomposed_20201120.pdf')
# img3 = pdf_to_image('HHT.pdf')
img3 = pdf_to_image('Rate_IP_IF.pdf')
img4 = pdf_to_image('QPO_profile.pdf')

# 创建2x2的子图布局
fig, axs = plt.subplots(2, 2, figsize=(10, 10))

# 定义标签
labels = ['a', 'b', 'c', 'd']

# 逐个子图添加图片和标签
for i, ax in enumerate(axs.flat):
    # 获取对应的图像
    img = [img1, img2, img3, img4][i]
    
    # 在子图上展示图像
    ax.imshow(np.array(img), aspect='auto')
    
    # 去除坐标轴
    ax.axis('off')
    
    # 在左上角添加标签
    ax.text(0.01, 1.02, labels[i], transform=ax.transAxes, fontsize=16, fontweight='bold', va='top', ha='left')

# 调整布局，避免子图重叠
plt.tight_layout()

# 保存为一个合并后的高分辨率PDF文件，设置dpi
plt.savefig('combined_images_high_res.pdf', dpi=300)  # 设置为300 DPI

# 显示图像
plt.show()
```
